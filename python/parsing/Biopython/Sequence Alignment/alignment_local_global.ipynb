```python
from Bio.Align import PairwiseAligner

seq1 = "AAATTTGGGCCCAAATTT"
seq2 = "TTTGAGCCC"

aligner = PairwiseAligner()
aligner.mode = "local"
aligner.match_score = 2
aligner.mismatch_score = -1
aligner.open_gap_score = -2
aligner.extend_gap_score = -0.5

alignments = aligner.align(seq1,seq2)
print(alignments[0])
print(alignments.score)

```

```python
def aligner(mode,match,missmatch,open_gap,extend_gap):
    aligner = PairwiseAligner()
    aligner.mode = mode
    aligner.match_score = match
    aligner.mismatch_score = missmatch
    aligner.open_gap_score = open_gap
    aligner.extend_gap_score = extend_gap
    return aligner
```

```python
aligner_global = aligner('global',2,-1,-1,-0.5)
aligner_global
```

```python
align_global = aligner_global.align(seq1,seq2)
```

```python
print(align_global.score)
print(align_global.sequences)
```
